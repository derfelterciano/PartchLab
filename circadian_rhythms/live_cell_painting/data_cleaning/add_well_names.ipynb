# We will add well names to the datasets here

```python
import polars as pl
import os
from pathlib import Path
import duckdb as db

import pandas as pd
```

```python
in_path = Path('/Users/dterciano/Desktop/PartchLabFiles/CircadianRhythms/Live Study/formatted_raw_datasets')

for file in in_path.iterdir():
    lf = pl.scan_csv(file, separator="\t")
    
    result = lf.with_columns(
        (
            pl.col("Row").map_elements(lambda x: chr(64 + x), return_dtype=pl.String) +
            pl.col("Column").cast(pl.String)
        ).alias("WellName")
    )
    
    print(file)
    display(result.select(["Row", "Column", "WellName"]).unique().sort(["Row", "Column"]).collect())
```
