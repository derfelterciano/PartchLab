# Circadian Cell Painting - Wavelet Analysis

We will be trying out the wavelet analysis module in our package

```python
import polars as pl
from pathlib import Path

from circadian_cell_painting.gene_expression import WaveletAnalyzer, WaveletResults, merge_data_platemap
from circadian_cell_painting.utils import retrieve_wells



detrended_dataset = Path("/home/derfelt/PartchLabFiles/Circadian Rhythms/Live "
                        "Study/enVision/enVision_preprocessed/2026-06-01._.20"
                        "xREP._.PREPROCESSED.parquet")

# plate_map = (
#     pl.read_excel("/home/derfelt/PartchLabFiles/Circadian Rhythms/Live "
#                   "Study/platemaps/2026_05_26_JUNE-EXP_plate_map.xlsx", 
#                   engine='openpyxl')
# )

plate_map = Path("/home/derfelt/PartchLabFiles/Circadian Rhythms/Live "
                  "Study/platemaps/2026_05_26_JUNE-EXP_plate_map.xlsx")

# merge datasets
dataset = merge_data_platemap(
    detrended_data=detrended_dataset, 
    plate_map=plate_map, 
    on_well_col="Well",
    gene_expression_feature='CalcResultI',
    )

out_dir = Path("/home/derfelt/PartchLabFiles/Circadian Rhythms/Live "
               "Study/enVision/TEST_WAVELET")
out_dir.mkdir(parents=True, exist_ok=True)

wells = retrieve_wells(data=dataset, well_col='Well')
print(wells)

display(dataset.filter(pl.col("Well") == "F09"))
wa = WaveletAnalyzer(
    periods=(16,50),
    sampling_interval=0.5,
    tick_interval=6
)

res = wa.analyze(
    dataset=dataset, 
    name_col='group_label',
    well_list=wells,
    well_col='Well',
    stat_val='median',
    timepoint_col='Timepoint'
    )

wa.visualize_all(
    wavelet_results=res,
    dataset=dataset,
    well_list=wells,
    well_col='Well',
    name_col='group_label',
    stat_val='median',
    timepoint_col="Timepoint",
    output_file=out_dir / "well_vis.pdf"
)

metrics = ['periods', 'amplitude', 'phase_hours', 'phase']
wa.interactive_ridges(
    wavelet_res=res,
    name_col='group_label',
    timepoint_col='time',
    metrics=metrics,
    out_file=out_dir / "interactive.html"
)
```
