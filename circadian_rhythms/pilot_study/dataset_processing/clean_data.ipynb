# There are some problesm with certain wells so we have to get rid of them

Remove the following:

- ~~Columns: 1-16~~
- Rows: B, D, F, H, J, L, N, P (REMOVE THE ENTIRE)

**ALSO REMOVE:**
- Columns 1 and 24 *ENTIRELY*

```python
import pandas as pd
import os
```

```python
all_rows = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J', 'K', 'L', 'M', 'N', 'O', 'P']
col_1_24 = [f"{row}{col}" for row in all_rows for col in [1, 24]]

remove_rows = ['B', 'D', 'F', 'H', 'J', 'L', 'N', 'P']
remove_cols = list(range(1,17))
row_remove_indices = [f"{row}{col}" for row in remove_rows for col in remove_cols]

remove_indices = list(set(row_remove_indices + col_1_24))

# remove_indices
```

```python
print(remove_indices)
```

Summary File processing

```python
summary_file_dir = "/home/derfelt/PartchLabFiles/Circadian Rhythms/Pilot Study/raw_data/Summaries"

exclude_cols = ['ScreenName', 'ScreenID', 'PlateName', 'PlateID', 'MeasurementDate',
       'MeasurementID', 'Row', 'Column', 'Timepoint',
       'Number of Analyzed Fields', 'Global Image Binning']

clean_file_dir = "/home/derfelt/PartchLabFiles/Circadian Rhythms/Pilot Study/cleaned_data/summaries_cleaned"

for file in os.listdir(summary_file_dir):
    # print(file)
    
    raw_out_name = os.path.splitext(file)[0]
    raw_out_name = f"{raw_out_name}_WELLS-DROPPED.tsv"
    out_file = os.path.join(clean_file_dir, raw_out_name)
    
    curr_file_path = os.path.join(summary_file_dir, file)
    print(curr_file_path)
    
    curr_df = pd.read_csv(curr_file_path, sep="\t", usecols= lambda col: col not in exclude_cols)
    curr_df.set_index('WellName', inplace=True)
    curr_df.drop(remove_indices, errors='ignore', inplace=True)
    
    curr_df.columns = curr_df.columns.str.strip()
    
    print(out_file)
    curr_df.to_csv(out_file, sep="\t")
```

HistDiff File Processing

```python
clean_file_dir = "/home/derfelt/PartchLabFiles/Circadian Rhythms/Pilot Study/cleaned_data/HistDiff_cleaned"
hd_file_dir = "/home/derfelt/PartchLabFiles/Circadian Rhythms/Pilot Study/raw_data/HistDiff Outputs"

exclude_cols = ["Row", "Timepoint", "Column"]

for file in os.listdir(hd_file_dir):
    curr_file = os.path.join(hd_file_dir, file)
    print(curr_file)
    
    raw_out = os.path.splitext(file)[0]
    raw_out = f"{raw_out}_WELL-DROPPED.csv"
    out_file = os.path.join(clean_file_dir, raw_out)
    print(out_file)
    
    curr_df = pd.read_csv(curr_file, usecols= lambda cols: cols not in exclude_cols, encoding='latin-1')
    curr_df.set_index("id", inplace=True)
    curr_df.drop(remove_indices, inplace=True)
    
    curr_df.columns = curr_df.columns.str.strip()
    curr_df.columns = curr_df.columns.str.replace("µ", "u")
    curr_df.columns = curr_df.columns.str.replace("Â", "")
    curr_df.columns = curr_df.columns.str.replace("²", "2")
    
    print(curr_df.shape)
    curr_df.to_csv(out_file, encoding='utf-8')
    
```
