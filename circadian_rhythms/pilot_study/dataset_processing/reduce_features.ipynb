```python
import pandas as pd
import os
import sys
```

```python
# features_list = '/Users/dterciano/Desktop/PartchLabFiles/CircadianRhythms/PilotStudy/Feature Reduction/results_full_0.85.txt'
features_list = '/home/derfelt/PartchLabFiles/Circadian Rhythms/Pilot Study/Feature_Reduction_Redone/redone_feat_reduction_full_0.9.txt'

feats = []
with open(features_list, 'r') as f:
    feats = [line.strip() for line in f]

print(len(feats))
```

```python
# out_dir = "/Users/dterciano/Desktop/PartchLabFiles/CircadianRhythms/PilotStudy/featRed_HistDiff_outs"
out_dir = "/home/derfelt/PartchLabFiles/Circadian Rhythms/Pilot Study/cleaned_data/feature_reduction_cleaned"

# in_dir = "/Users/dterciano/Desktop/PartchLabFiles/CircadianRhythms/PilotStudy/HistDiff Outputs"
in_dir = "/home/derfelt/PartchLabFiles/Circadian Rhythms/Pilot Study/cleaned_data/HistDiff_cleaned"

for file in os.listdir(in_dir):
    curr_file_path = os.path.join(in_dir, file)
    
    print(curr_file_path)
    curr_df = pd.read_csv(curr_file_path, index_col=0)
    curr_df.columns = curr_df.columns.str.strip()
    
    curr_df = curr_df[feats]
    
    out_file_name = f"{os.path.splitext(file)[0]}_FeatRed.csv"
    out_file_path = os.path.join(out_dir, out_file_name)
    
    curr_df.to_csv(out_file_path)
    print(curr_df.shape)

    
```
