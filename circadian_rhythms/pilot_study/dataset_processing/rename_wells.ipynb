# We need to rename id cols to the MoleculeID names

```python
import pandas as pd
import os
import re
```

```python
platemap = "/home/derfelt/PartchLabFiles/Circadian Rhythms/Pilot Study/2025-06-19transfer_map.xlsx"

platemap = pd.read_excel(platemap)
platemap["human_readable_ids"] = platemap["MoleculeID"].astype(str) + "._." + platemap["384_Well"].astype(str)
platemap["384_Well"] = platemap["384_Well"].str.replace(r'(?<=\D)0+', "", regex=True)
platemap = platemap[["384_Well", "human_readable_ids"]]
display(platemap)
```

```python
feat_red_hd_path = "/home/derfelt/PartchLabFiles/Circadian Rhythms/Pilot Study/cleaned_data/feature_reduction_cleaned"
out_path = "/home/derfelt/PartchLabFiles/Circadian Rhythms/Pilot Study/cleaned_data/feature_reduction_cleaned_renamed"
for file in os.listdir(feat_red_hd_path):
    timepoint = re.search(r'(\d+)h', file).group(0)
    condition = re.search(r'\d+h_(\w+?)_', file).group(1)
    print(timepoint, condition, file)
    
    curr_file = os.path.join(feat_red_hd_path, file)
    curr_df = pd.read_csv(curr_file)
    print(curr_df.shape)
    
    merge_df = pd.merge(left=curr_df, right=platemap, left_on="id", right_on="384_Well", how="left")
    
    mask = merge_df["384_Well"].isna()
    merge_df.loc[mask, "human_readable_ids"] = "NO-TREATMENT._." + merge_df.loc[mask, "id"]
    merge_df["human_readable_ids"] = merge_df["human_readable_ids"].astype(str) + "._." + timepoint + "._." + condition
    
    merge_df.drop(['384_Well', 'id'], inplace=True, axis=1)
    merge_df.set_index("human_readable_ids", inplace=True)
    display(merge_df)
    merge_df.to_csv(os.path.join(out_path, file))
```
