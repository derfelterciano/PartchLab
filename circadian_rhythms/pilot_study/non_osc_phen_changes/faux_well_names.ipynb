# lets replace the real data with the faux names and then process it

```python
import pandas as pd
from pathlib import Path

cell_data = Path("/Users/dterciano/Desktop/PartchLabFiles/CircadianRhythms/PilotStudy/non_oscilating_data/no_treatment_cell_data.csv")
faux_pm = Path("/Users/dterciano/Desktop/PartchLabFiles/CircadianRhythms/PilotStudy/non_oscilating_data/faux_platemap.csv")
faux_df = pd.read_csv(faux_pm)
```

```python
cell_df = pd.read_csv(cell_data)

df = pd.merge(left=cell_df, right=faux_df, left_on="WellName", right_on="formatted_name", how="left")
df.drop(columns=['WellName', 'formatted_name'], inplace=True)
df.rename(columns={'faux_name': 'WellName'}, inplace=True)

df.to_csv("/Users/dterciano/Desktop/PartchLabFiles/CircadianRhythms/PilotStudy/non_oscilating_data/faux_cell_data.txt", sep="\t", index=False)
display(df)
```

# Rename the processed HD files back to normal names

```python
hd_df = Path("/Users/dterciano/Desktop/PartchLabFiles/CircadianRhythms/PilotStudy/non_oscilating_data/no_treatment_HD.csv")
hd_df = pd.read_csv(hd_df)

df = pd.merge(left=hd_df, left_on="id", right=faux_df, right_on="faux_name", how="left")
df.drop(columns=["faux_name", "id"], inplace=True)
df.rename(columns={"formatted_name": "WellName"}, inplace=True)
df.set_index("WellName", inplace=True)

df.to_csv("/Users/dterciano/Desktop/PartchLabFiles/CircadianRhythms/PilotStudy/non_oscilating_data/no_treatment_HD_renamed.csv")

display(df)
```
