# Reduce features for dataset

```python
import pandas as pd
from pathlib import Path

raw_hd = Path('/home/derfelt/PartchLabFiles/Circadian Rhythms/Pilot Study/no_treatmnet_data/no_treatment_HD_renamed.csv')
```

```python
features_list = Path('/home/derfelt/PartchLabFiles/Circadian Rhythms/Pilot Study/Feature_Reduction_Redone/redone_feat_reduction_full_0.9.txt')
feats = []

with open(features_list, "r") as f:
    feats = [line.strip() for line in f]

print(len(feats))
```

```python
exclude_cols = ['Column', 'Row', 'Timepoint']
hd = pd.read_csv(raw_hd, index_col=0, usecols=lambda col: col not in exclude_cols)

hd.columns = hd.columns.str.strip()
hd.columns = hd.columns.str.replace("µ", "u")
hd.columns = hd.columns.str.replace("Â", "")
hd.columns = hd.columns.str.replace("²", "2")
hd = hd[feats]

hd.to_csv("/home/derfelt/PartchLabFiles/Circadian Rhythms/Pilot Study/no_treatmnet_data/no-treatment_hd_feat_red.csv")

display(hd)
```
