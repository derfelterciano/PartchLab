# Phenotypic changes in non-oscilating cells per time

## In this workbook: maniplating and combing data

```python
import pandas as pd
from pathlib import Path
import duckdb

timepoint_dict = {
    "250619_163847-V_cellByCell_data.tsv" : "24h",
    "250619_211820-V_cellByCell_data.tsv" : "66h",
    "250620_015806-V_cellByCell_data.tsv" : "72h"
}

raw_cell_data = "/Users/dterciano/Desktop/PartchLabFiles/CircadianRhythms/PilotStudy/raw_cell_data"
raw_cell_data = Path(raw_cell_data)

for path in raw_cell_data.iterdir():
    if not path.name.startswith("."):
        print(path)
```

```python
test_df = pd.read_csv(next(raw_cell_data.iterdir()), nrows=0, sep="\t")
print(*test_df.columns.tolist(), sep="\n")
```

```python
dfs = []
for path in raw_cell_data.iterdir():
    if not path.name.startswith("."):
        print(path.name)
        relation = duckdb.read_csv(path, delimiter='\t', strict_mode=False, ignore_errors=True)
        df = duckdb.sql("""
            SELECT * FROM relation
            WHERE WellName LIKE 'A%' OR WellName LIKE 'I%'
        """).df()
        
        print(timepoint_dict[path.name])
        
        df['WellName'] = df["WellName"] + "._." + timepoint_dict[path.name]
        
        dfs.append(df)

no_treatments = pd.concat(dfs, axis=0)
display(no_treatments)
```

```python
no_treatments.to_csv("/Users/dterciano/Desktop/PartchLabFiles/CircadianRhythms/PilotStudy/non_oscilating_data/no_treatment_cell_data.csv", index=False)
```

# Generate Faux platemap

```python
import pandas as pd
import itertools

timepoints = ['24h', '66h', '72h']
original_rows = ['A', 'I']
columns = range(1, 25)
faux_rows = 'ABCDEF'

mapping = []
faux_row_iter = iter(faux_rows)

for tp in timepoints:
    for orig_row in original_rows:
        faux_row = next(faux_row_iter)
        for col in columns:
            formatted = f"{orig_row}{col}._.{tp}"
            faux = f"{faux_row}{col}"
            mapping.append((formatted, faux))

platemap = pd.DataFrame(mapping, columns=['formatted_name', 'faux_name'])
platemap.to_csv('/Users/dterciano/Desktop/PartchLabFiles/CircadianRhythms/PilotStudy/non_oscilating_data/faux_platemap.csv', index=False)
display(platemap)
```
